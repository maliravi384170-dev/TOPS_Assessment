```python
class ClinicAppointment:
    def __init__(self):
        # Available time slots
        self.time_slots = ["10am", "11am", "12pm", "2pm", "3pm"]
        
        # Appointments stored in memory
        # Structure:
        # {
        #   doctor_name: {
        #       time_slot: [appointment_dict, appointment_dict]
        #   }
        # }
        self.appointments = {}

    def book_appointment(self, name, age, mobile, doctor):
        print("\nAvailable Time Slots:")
        for i, slot in enumerate(self.time_slots, 1):
            print(f"{i}. {slot}")

        choice = int(input("Choose slot number: ")) - 1
        if choice < 0 or choice >= len(self.time_slots):
            print("❌ Invalid slot selection")
            return

        slot = self.time_slots[choice]

        # Initialize doctor if not present
        if doctor not in self.appointments:
            self.appointments[doctor] = {}

        # Initialize slot if not present
        if slot not in self.appointments[doctor]:
            self.appointments[doctor][slot] = []

        # Check slot limit (max 3)
        if len(self.appointments[doctor][slot]) >= 3:
            print("❌ Slot full. Please choose another time.")
            return

        # Store appointment
        appointment = {
            "name": name,
            "age": age,
            "mobile": mobile,
            "doctor": doctor,
            "slot": slot
        }

        self.appointments[doctor][slot].append(appointment)
        print("✅ Appointment booked successfully!")

    def view_appointment(self, mobile):
        for doctor in self.appointments:
            for slot in self.appointments[doctor]:
                for appt in self.appointments[doctor][slot]:
                    if appt["mobile"] == mobile:
                        print("\n📋 Appointment Details")
                        print(f"Name   : {appt['name']}")
                        print(f"Age    : {appt['age']}")
                        print(f"Doctor : {appt['doctor']}")
                        print(f"Time   : {appt['slot']}")
                        return
        print("❌ No appointment found")

    def cancel_appointment(self, mobile):
        for doctor in self.appointments:
            for slot in self.appointments[doctor]:
                for appt in self.appointments[doctor][slot]:
                    if appt["mobile"] == mobile:
                        self.appointments[doctor][slot].remove(appt)
                        print("✅ Appointment cancelled successfully")
                        return
        print("❌ No appointment found to cancel")
```

```python
class SchoolManagement:
    def __init__(self):
        self.students = {}   # Stores student records
        self.next_id = 1001  # Auto-increment Student ID

    # ---------------- New Admission ----------------
    def new_admission(self, name, age, student_class, mobile):
        # Age validation
        if age < 5 or age > 18:
            print("❌ Age must be between 5 and 18")
            return

        # Class validation
        if student_class < 1 or student_class > 12:
            print("❌ Class must be between 1 and 12")
            return

        # Mobile validation
        if not (mobile.isdigit() and len(mobile) == 10):
            print("❌ Mobile number must be 10 digits")
            return

        student_id = self.next_id
        self.students[student_id] = {
            "name": name,
            "age": age,
            "class": student_class,
            "mobile": mobile
        }

        self.next_id += 1
        print(f"✅ Admission Successful! Student ID: {student_id}")

    # ---------------- View Student ----------------
    def view_student(self, student_id):
        if student_id in self.students:
            s = self.students[student_id]
            print("\n📋 Student Details")
            print(f"ID     : {student_id}")
            print(f"Name   : {s['name']}")
            print(f"Age    : {s['age']}")
            print(f"Class  : {s['class']}")
            print(f"Mobile : {s['mobile']}")
        else:
            print("❌ Student ID not found")

    # ---------------- Update Student ----------------
    def update_student(self, student_id):
        if student_id not in self.students:
            print("❌ Student ID not found")
            return

        print("\n1. Update Mobile Number")
        print("2. Update Class")
        choice = input("Choose option: ")

        if choice == "1":
            new_mobile = input("Enter new mobile number: ")
            if new_mobile.isdigit() and len(new_mobile) == 10:
                self.students[student_id]["mobile"] = new_mobile
                print("✅ Mobile number updated")
            else:
                print("❌ Invalid mobile number")

        elif choice == "2":
            new_class = int(input("Enter new class (1–12): "))
            if 1 <= new_class <= 12:
                self.students[student_id]["class"] = new_class
                print("✅ Class updated")
            else:
                print("❌ Invalid class")

        else:
            print("❌ Invalid choice")

    # ---------------- Remove Student ----------------
    def remove_student(self, student_id):
        if student_id in self.students:
            del self.students[student_id]
            print("✅ Student record removed")
        else:
            print("❌ Student ID not found")
```

```python
class BusReservation:
    def __init__(self):
        # Predefined routes with prices
        self.routes = {
            1: {"route": "Mumbai to Pune", "price": 500},
            2: {"route": "Delhi to Jaipur", "price": 600},
            3: {"route": "Ahmedabad to Surat", "price": 400},
            4: {"route": "Bangalore to Chennai", "price": 700}
        }

        self.tickets = {}          # ticket_id -> ticket details
        self.route_seats = {}      # route_name -> booked seats
        self.next_ticket_id = 10001
        self.max_seats = 40

    # ---------------- Show Routes ----------------
    def show_routes(self):
        print("\n🚌 Available Routes")
        for key, value in self.routes.items():
            print(f"{key}. {value['route']} - ₹{value['price']}")

    # ---------------- Book Ticket ----------------
    def book_ticket(self, name, age, mobile, route_choice):
        if route_choice not in self.routes:
            print("❌ Invalid route selection")
            return

        if not (mobile.isdigit() and len(mobile) == 10):
            print("❌ Invalid mobile number")
            return

        route_name = self.routes[route_choice]["route"]
        price = self.routes[route_choice]["price"]

        # Initialize seat tracking
        if route_name not in self.route_seats:
            self.route_seats[route_name] = []

        # Check seat availability
        if len(self.route_seats[route_name]) >= self.max_seats:
            print("❌ Bus is full for this route")
            return

        seat_number = len(self.route_seats[route_name]) + 1
        ticket_id = self.next_ticket_id

        ticket = {
            "name": name,
            "age": age,
            "mobile": mobile,
            "route": route_name,
            "price": price,
            "seat": seat_number
        }

        self.tickets[ticket_id] = ticket
        self.route_seats[route_name].append(seat_number)
        self.next_ticket_id += 1

        print("✅ Ticket Booked Successfully!")
        print(f"🎟 Ticket ID : {ticket_id}")
        print(f"🪑 Seat No   : {seat_number}")
        print(f"💰 Fare      : ₹{price}")

    # ---------------- View Ticket ----------------
    def view_ticket(self, ticket_id):
        if ticket_id in self.tickets:
            t = self.tickets[ticket_id]
            print("\n🎟 Ticket Details")
            print(f"Ticket ID : {ticket_id}")
            print(f"Name      : {t['name']}")
            print(f"Age       : {t['age']}")
            print(f"Mobile    : {t['mobile']}")
            print(f"Route     : {t['route']}")
            print(f"Seat No   : {t['seat']}")
            print(f"Fare      : ₹{t['price']}")
        else:
            print("❌ Ticket not found")

    # ---------------- Cancel Ticket ----------------
    def cancel_ticket(self, ticket_id):
        if ticket_id in self.tickets:
            route = self.tickets[ticket_id]["route"]
            seat = self.tickets[ticket_id]["seat"]
            self.route_seats[route].remove(seat)
            del self.tickets[ticket_id]
            print("✅ Ticket cancelled successfully")
        else:
            print("❌ Ticket ID not found")
```
